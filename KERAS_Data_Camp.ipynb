### **Как определить количество скрытых слоёв и нейронов в модели?**
Выбор количества слоёв и нейронов — это **эмпирический процесс**, зависящий от сложности задачи, объёма данных и требований к точности модели. Давай разберём основные принципы, которые помогут сделать правильный выбор.

---

## **1. Количество скрытых слоёв**
📌 **Общее правило**:
- **Простые задачи** (линейные или почти линейные зависимости) → **0–1 скрытый слой**.
- **Умеренные задачи** (полиномиальные зависимости, нелинейные зависимости) → **1–3 скрытых слоя**.
- **Сложные задачи** (изображения, текст, сложные зависимости) → **4+ слоёв (глубокие нейросети)**.

🔹 **Когда хватает одного скрытого слоя?**
- Если задача **простая** (например, линейная или квадратичная зависимость).
- Если модель **работает хорошо** уже с одним слоем.

🔹 **Когда нужно больше слоёв?**
- Если модель **плохо обучается** с одним слоем.
- Если зависимость **сложная и нелинейная** (например, многомерные функции или задачи классификации изображений).
- Если добавление нейронов в одном слое **не даёт прироста качества**.

👉 **В нашем случае (траектория метеорита) у нас квадратичная зависимость, поэтому 2-3 слоя уже достаточно, чтобы уловить кривизну траектории.**

---

## **2. Количество нейронов в слоях**
📌 **Как выбрать количество нейронов в скрытых слоях?**
1. **Количество нейронов в первом слое** обычно **больше, чем количество входных признаков** (обычно 2x или 3x от входного размера).
2. **Чем сложнее функция, тем больше нейронов может понадобиться.**
3. **Начинай с 16–128 нейронов на слой и экспериментируй.**
4. **Чем больше данных — тем больше нейронов можно использовать без риска переобучения.**
5. **Если данных мало, большое число нейронов может привести к переобучению.**

### **Практические рекомендации по выбору нейронов**
| Количество входных признаков | Рекомендуемое число нейронов в скрытых слоях |
|-----------------------------|--------------------------------------|
| 1-10                        | 16–64                               |
| 10-100                      | 64–256                              |
| 100+                        | 256+ (но с регуляризацией)         |

🔹 **Когда мало нейронов — модель недообучается** (не может уловить сложные зависимости).  
🔹 **Когда слишком много нейронов — модель переобучается** (запоминает данные, но плохо работает на новых примерах).  

### **Почему мы выбрали 50 нейронов в модели с метеоритом?**
- **Входной размер — 1 (время)**, но зависимость **квадратичная** → значит, модель должна уловить нелинейность.
- **50 нейронов в каждом слое** — достаточно для моделирования квадратичной зависимости, но не слишком много, чтобы переобучиться.
- **ReLU как функция активации** помогает выделять важные признаки и сохранять градиенты.

---

## **3. Подход к настройке количества слоёв и нейронов**
### 🔍 **Методы подбора структуры модели**
1. **Начни с малого**
   - 1-2 скрытых слоя с 16-64 нейронами.
   - Обучи модель и оцени качество.
   
2. **Постепенно увеличивай сложность**
   - Добавляй нейроны (например, увеличь с 16 → 32 → 64).
   - Добавляй скрытые слои (например, с 1 → 2 → 3).

3. **Следи за переобучением**
   - Если ошибка на тренировке **сильно меньше**, чем на тесте → слишком много нейронов/слоёв.
   - Пробуй **добавлять dropout или L2-регуляризацию**.

4. **Используй кросс-валидацию**
   - Раздели данные на **тренировочный и валидационный** наборы.
   - Оцени, как модель ведёт себя на новых данных.

5. **Grid Search и Random Search**
   - Используй `GridSearchCV` или `RandomizedSearchCV` (из библиотеки `scikit-learn`) для автоматического подбора гиперпараметров.

---

## **4. Пример экспериментов**
Допустим, у нас есть задача предсказать траекторию метеорита. Мы можем протестировать разные архитектуры:

| Архитектура | Количество слоёв | Количество нейронов | Loss (ошибка) |
|------------|---------------|-----------------|--------------|
| 1 скрытый слой, 10 нейронов | 1 | 10 | 0.50 |
| 1 скрытый слой, 50 нейронов | 1 | 50 | 0.20 |
| 2 скрытых слоя, 50 нейронов | 2 | 50 | 0.12 |
| 3 скрытых слоя, 50 нейронов | 3 | 50 | 0.11 |

👉 **Вывод:** 2-3 слоя по 50 нейронов — хорошее компромиссное решение.

✅ **Для сложных задач:** 3+ слоёв, 128+ нейронов, но с регуляризацией.

### **Почему количество нейронов в слоях часто кратно 16?**  

Это **не жёсткое правило**, но есть несколько **технических и эмпирических причин**, почему числа вроде **16, 32, 64, 128, 256** часто используются в нейронных сетях.

---

## **1. Аппаратные ограничения (оптимизация для GPU)**
🚀 **Современные графические процессоры (GPU) обрабатывают данные блоками (warps, threads, SIMD)**.  
🔹 GPU работают **оптимально**, когда операции выполняются с размерами, кратными **16, 32, 64, 128 и т. д.**  
🔹 Если число нейронов **не кратно этим значениям**, то могут возникать **потери производительности** из-за "разброса" вычислений по разным потокам.

✅ **Пример:**  
- Допустим, у нас **25 нейронов в слое** → GPU может загружать их **в два блока по 16 и 9 нейронов**, но второй блок будет работать неэффективно.  
- Если взять **32 нейрона**, то вся матрица обрабатывается без пустых вычислительных ячеек.

---

## **2. Эффективность работы с матрицами (BLAS, CUDA, Tensor Cores)**
💡 **В машинном обучении много матричных операций (умножение, свёртка и т. д.).**  
🔹 Библиотеки, которые выполняют эти операции (например, **cuBLAS, OpenBLAS, MKL**), **предпочитают размеры, кратные степеням двойки** (16, 32, 64 и т. д.).  
🔹 Это связано с тем, что память и регистры оптимизированы для таких размеров, и это снижает накладные расходы.

✅ **Пример:**  
- GPU (например, NVIDIA Tensor Cores) работают с матрицами **размера 16×16** → значит, если у нас **16, 32, 64 нейрона**, то мы получим максимальную производительность.

---

## **3. Быстрая обработка и кэширование в CPU/GPU**
🔹 Процессоры и видеокарты используют **кэш-память** для хранения промежуточных данных.  
🔹 Если размеры данных **кратны степеням двойки (16, 32, 64 и т. д.)**, то кэш **заполняется полностью**, а если не кратны — могут быть **лишние обращения к памяти (более медленные операции)**.  
🔹 Это особенно важно для **батчей данных** в нейронных сетях.

✅ **Пример:**  
- Если у нас **30 нейронов**, но процессор оптимизирован под **32**, он всё равно выделит место на **32**, и 2 ячейки будут **пустыми (неэффективное использование памяти).**

---

## **4. Эмпирическое правило в глубоком обучении**
🔹 **Исследования показали**, что модели с нейронами кратными **16, 32, 64** **чаще** дают **лучшие результаты по скорости и точности**.  
🔹 Особенно это заметно при работе с **большими архитектурами**, такими как **ResNet, VGG, Transformer, GPT**, где число каналов, фильтров и нейронов **кратно 16, 32, 64 и т. д.**

✅ **Пример:**  
- В сверточных нейросетях (CNN) часто используют **64, 128, 256 каналов**, потому что так **удобно работать с изображениями**.

---

## **5. Исключения: можно ли использовать не кратные 16 числа?**
Конечно! 😃  
🔹 **Если модель маленькая**, разница в производительности будет **незаметна**.  
🔹 **Иногда подбирают "нестандартное" количество нейронов** (например, 10, 25, 75) **на основе экспериментов**.  
🔹 Но **если у тебя мощный GPU, лучше использовать кратные 16, 32, 64** для максимальной скорости.

---

## **Вывод: стоит ли всегда выбирать кратные 16 нейронов?**
✅ **Да, если модель большая и работает на GPU — это ускорит обучение.**  
✅ **Да, если ты хочешь максимально эффективно использовать память и вычислительные ресурсы.**  
🤷 **Но если ты обучаешь маленькую модель на CPU — разница будет минимальной.**

**A multi-class model**
You're going to build a model that predicts who threw which dart only based on where that dart landed! (That is the dart's x and y coordinates on the board.)

This problem is a multi-class classification problem since each dart can only be thrown by one of 4 competitors. So classes/labels are mutually exclusive, and therefore we can build a neuron with as many output as competitors and use the softmax activation function to achieve a total sum of probabilities of 1 over all competitors.

The Sequential model and Dense layers are already imported for you to use.

````python
# Instantiate a sequential model
model = Sequential()

# Add 3 dense layers of 128, 64 and 32 neurons each
model.add(Dense(128, input_shape=(2,), activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))

# Add a dense layer with as many neurons as competitors
model.add(Dense(4, activation='softmax'))

# Compile your model using categorical_crossentropy loss
model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
````

Preparing a model for tuning
Let's tune the hyperparameters of a binary classification model that does well classifying the breast cancer dataset.

You've seen that the first step to turn a model into a sklearn estimator is to build a function that creates it. The definition of this function is important since hyperparameter tuning is carried out by varying the arguments your function receives.

Build a simple create_model() function that receives both a learning rate and an activation function as arguments. The Adam optimizer has been imported as an object from tensorflow.keras.optimizers so that you can also change its learning rate parameter.

````python
# Creates a model given an activation and learning rate
def create_model(learning_rate, activation):

  	# Create an Adam optimizer with the given learning rate
  	opt = Adam(lr = 'adam')

  	# Create your binary classification model
  	model = Sequential()
  	model.add(Dense(128, input_shape = (30,), activation = 'relu'))
  	model.add(Dense(256, activation = 'relu'))
  	model.add(Dense(1, activation = 'sigmoid'))

  	# Compile your model with your optimizer, loss, and metrics
  	model.compile(optimizer = opt, loss = 'binary cross-entropy', metrics = ['accuracy'])
  	return model
````

Building an autoencoder
Autoencoders have several interesting applications like anomaly detection or image denoising. They aim at producing an output identical to its inputs. The input will be compressed into a lower dimensional space, encoded. The model then learns to decode it back to its original form.

You will encode and decode the MNIST dataset of handwritten digits, the hidden layer will encode a 32-dimensional representation of the image, which originally consists of 784 pixels (28 x 28). The autoencoder will essentially learn to turn the 784 pixels original image into a compressed 32 pixels image and learn how to use that encoded representation to bring back the original 784 pixels image.

The Sequential model and Dense layers are ready for you to use.

Let's build an autoencoder!

````python
# Start with a sequential model
autoencoder = Sequential()

# Add a dense layer with input the original image pixels and neurons the encoded representation
autoencoder.add(Dense(32, input_shape=(784, ), activation="relu"))

# Add an output layer with as many neurons as the orginal image pixels
autoencoder.add(Dense(784, activation = "sigmoid"))

# Compile your model with adadelta
autoencoder.compile(optimizer = 'adadelta', loss = 'binary_crossentropy')

# Summarize your model structure
autoencoder.summary()
````

During the following exercises you will build a toy LSTM model that is able to predict the next word using a small text dataset. This dataset consist of cleaned quotes from the The Lord of the Ring movies. You can find them in the text variable.

You will turn this text into sequences of length 4 and make use of the Keras Tokenizer to prepare the features and labels for your model!

The Keras Tokenizer is already imported for you to use. It assigns a unique number to each unique word, and stores the mappings in a dictionary. This is important since the model deals with numbers but we later will want to decode the output numbers back into words.

````python
# Split text into an array of words
words = text.split()

# Make sentences of 4 words each, moving one word at a time
sentences = []
for i in range(4, len(words)):
  sentences.append(' '.join(words[i-4:i]))

# Instantiate a Tokenizer, then fit it on the sentences
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

# Turn sentences into a sequence of numbers
sequences = tokenizer.texts_to_sequences(sentences)
print("Sentences: \n {} \n Sequences: \n {}".format(sentences[:5],sequences[:5]))
````

Build your LSTM model
You've already prepared your sequences of text. It's time to build your LSTM model!

Remember your sequences had 4 words each, your model will be trained on the first three words of each sequence, predicting the 4th one. You are going to use an Embedding layer that will essentially learn to turn words into meaningful vectors. These vectors will then be passed to a simple LSTM layer. Our output is a Dense layer with as many neurons as words in the vocabulary and softmax activation. This is because we want to obtain the highest probable next word out of all possible words.

The size of the vocabulary of words (the unique number of words) is stored in vocab_size.

````python
# Import the Embedding, LSTM and Dense layer
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential()

# Add an Embedding layer with the right parameters
model.add(Embedding(input_dim = vocab_size, input_length = 3, output_dim = 8, ))

# Add a 32 unit LSTM layer
model.add(LSTM(32))

# Add a hidden Dense layer of 32 units and an output layer of vocab_size with softmax
model.add(Dense(32, activation='relu'))
model.add(Dense(vocab_size, activation='softmax'))
model.summary()
````

Decode your predictions
Your LSTM model has already been trained (details in the previous exercise success message) so that you don't have to wait. It's time to define a function that decodes its predictions. The trained model will be passed as a default parameter to this function.

Since you are predicting on a model that uses the softmax function, numpy's argmax() can be used to obtain the index/position representing the most probable next word out of the output vector of probabilities.

The tokenizer you previously created and fitted, is loaded for you. You will be making use of its internal index_word dictionary to turn the model's next word prediction (which is an integer) into the actual written word it represents.

You're very close to experimenting with your model!

````python
def predict_text(test_text, model = model):
  if len(test_text.split()) != 3:
    print('Text input should be 3 words!')
    return False

  # Turn the test_text into a sequence of numbers
  test_seq = tokenizer.texts_to_sequences([test_text])
  test_seq = np.array(test_seq)

  # Use the model passed as a parameter to predict the next word
  pred = model.predict(test_seq).argmax(axis = 1)[0]

  # Return the word that maps to the prediction
  return tokenizer.index_word[pred]
````
